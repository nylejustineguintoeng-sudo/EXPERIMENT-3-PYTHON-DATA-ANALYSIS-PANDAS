### A. POSITIONAL AND LABEL-BASED SLICING

```python
import pandas as pd
# Load the CSV file into a DataFrame named cars
cars = pd.read_csv('cars.csv')
```

```python
# a. Display shape and complete list of column names
print("Cars dataset shape:", cars.shape)
print("Column names of cars:")
print(list(cars.columns))

# b. Positional slicing for rows 6 through 10 using .iloc
cars_6_to_10 = cars.iloc[5:10]
print("\ncars rows 6 to 10:")
print(cars_6_to_10)

# c. Display specified columns using column labels
selected_cars_6_to_10 = cars_6_to_10[['Model', 'mpg', 'cyl', 'hp', 'gear']]
print("\nselected rows 6 to 10:")
print(selected_cars_6_to_10)
```

### B. MODEL LOOKUP

```python
# a. Display complete row for Toyota Corolla
toyota = cars[cars['Model'] == 'Toyota Corolla']

# b. Display specified columns for Pontiac Firebird
pontiac = cars[cars['Model'] == 'Pontiac Firebird'][['Model', 'mpg', 'hp', 'wt']]

# Display outputs
print("Toyota Corolla:")
print(toyota)
print("\nPontiac Firebird:")
print(pontiac)
```

### C. MULTI-MODEL SUBSETTING

```python
# Create selected_cars DataFrame using boolean condition
selected_cars = cars[cars['Model'].isin(['Datsun 710', 'Lotus Europa', 'Ferrari Dino'])][['Model', 'mpg', 'cyl', 'hp', 'gear']]

# Display DataFrame and shape check
print("Selected Cars:")
print(selected_cars)
print("\nShape of selected_cars:", selected_cars.shape)
```
